2b-SQL Database operations

```python
import sqlite3
import pandas as pd
```

```python
conn=sqlite3.connect("Students.db")
cursor=conn.cursor()
```

```python
cursor.execute('''CREATE TABLE IF NOT EXISTS Students (id INTEGER,name TEXT)''')
```

```python
cursor.execute("INSERT INTO Students VALUES(1, 'John')")
```

```python
# Read directly into a DataFrame
df = pd.read_sql_query("SELECT * FROM Students", conn)
print(df)
```

```python
conn.commit()
```

```python
cursor.execute("INSERT INTO Students VALUES(2, 'Joy')")
conn.commit()
```

```python
cursor.execute("INSERT INTO Students VALUES(3, 'Alice')")
conn.commit()
```

```python
# Update the record in the table
cursor.execute("UPDATE Students SET name = 'Alice' WHERE id = 1")
conn.commit()
```

```python
# Delete a record from the table
cursor.execute("DELETE FROM Students WHERE id = 1")
conn.commit()
conn.close()
```

```python
# Reopen the database connection
conn = sqlite3.connect("Students.db")

# Read and print the table
df = pd.read_sql_query("SELECT * FROM Students", conn)
print(df)

```
