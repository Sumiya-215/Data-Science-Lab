```python
import pandas as pd
```

```python
index=[['Engineering','Engineering','Science','Science'],['CSE','ECE','Pyysics','Chemistry']]
```

```python
multi_index=pd.MultiIndex.from_arrays(index,names=['Department','Branch'])
```

```python
marks = pd.Series(
      [85, 78, 92, 88],
          index=multi_index
          )
```

```python
print("Original Series:")
print(marks)
```

```python
print("\nData for Engineering:")
print(marks.loc['Engineering'])
```

```python
print("\nData for CSE:")
print(marks.loc[('Engineering', 'CSE')])
```

```python
print("\nData for Science:")
print(marks.loc['Science'])
```
