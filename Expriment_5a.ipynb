```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 180, 160, 220, 250]
expenses = [80, 100, 120, 110, 140, 160]
```

```python
fig, ax = plt.subplots(2, 1, figsize=(10, 8))
```

```python
ax[0].plot(
      months,
          sales,
              marker="o",
                  linewidth=2,
                      label="Sales"
                      )

ax[0].set_title("Monthly Sales")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")
```

```python
ax[0].set_xticks(range(len(months)))
ax[0].set_xticklabels(months)
```

```python
ax[0].annotate(
      "Highest Sales",
          xy=(5, 250),
              xytext=(3.5, 270),
                  arrowprops=dict(arrowstyle="->")
                  )

ax[0].legend()
ax[0].grid(True)
```

```python
ax[1].plot(
      months,
          expenses,
              marker="s",
                  linewidth=2,
                      label="Expenses"
                      )

ax[1].set_title("Monthly Expenses")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Expenses")
```

```python
ax[1].set_xticks(range(len(months)))
ax[1].set_xticklabels(months)
```

```python
ax[1].annotate(
      "Highest Expense",
          xy=(5, 160),
              xytext=(3.5, 180),
                  arrowprops=dict(arrowstyle="->")
                  )

ax[1].legend()
ax[1].grid(True)
```

```python
fig.suptitle("Monthly Sales and Expenses Analysis", fontsize=16)
```

```python
plt.tight_layout()
```

```python
plt.savefig(
      "monthly_sales_expenses.png",
          dpi=300,
              bbox_inches="tight"
              )

              # Display the plot
plt.show()
```
